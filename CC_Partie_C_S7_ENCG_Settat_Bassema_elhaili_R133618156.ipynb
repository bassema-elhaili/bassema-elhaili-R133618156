```python
# Sources réelles intégrées à scikit-learn : trois problèmes différents
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes, load_breast_cancer, load_wine
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, f1_score, roc_auc_score, silhouette_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
GRAINE = 42

jeu_regression = load_diabetes()
X_reg, y_reg = jeu_regression.data, jeu_regression.target
jeu_classification = load_breast_cancer()
X_cls, y_cls = jeu_classification.data, jeu_classification.target
jeu_non_supervise = load_wine()
X_clu = StandardScaler().fit_transform(jeu_non_supervise.data)  # Exploration sans cible

Xr_ent, Xr_test, yr_ent, yr_test = train_test_split(X_reg, y_reg, test_size=.25, random_state=GRAINE)
Xc_ent, Xc_test, yc_ent, yc_test = train_test_split(X_cls, y_cls, test_size=.25, stratify=y_cls, random_state=GRAINE)
print('Diabetes:', X_reg.shape, '| Breast Cancer:', X_cls.shape, '| Wine:', X_clu.shape)
print('Ces données sont réelles ; les étiquettes du vin ne sont pas employées dans le clustering.')
```

## 1. Régression linéaire

**Variante C — question spécifique :** Quel diagnostic du résidu indique une mauvaise spécification linéaire ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.linear_model import LinearRegression
modele = LinearRegression().fit(Xr_ent, yr_ent)
predictions = modele.predict(Xr_test)
print('MAE régression linéaire :', round(mean_absolute_error(yr_test, predictions), 2))
print('R² :', round(r2_score(yr_test, predictions), 3))
```

### Présentation (C1)
* **Catégorie :** Apprentissage supervisé (régression linéaire paramétrique).
* **Nature des entrées / sorties :** En entrée, une matrice de variables explicatives $X$ ; en sortie, une variable cible continue $y$.
* **Hypothèse clé :** La relation entre les variables explicatives et la cible est supposée linéaire.
* **Étapes successives de calcul :** (1) Initialisation du vecteur des paramètres, (2) Application de la méthode des moindres carrés ordinaires (OLS) pour minimiser la somme des carrés des résidus, (3) Calcul des coefficients optimaux.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les coefficients (poids $\beta$) et l'ordonnée à l'origine (biais $\beta_0$) ; la régression linéaire simple ne comporte pas d'hyperparamètre d'optimisation majeur.
* **Équation et définition des symboles :**
$$\hat{y}_i = \beta_0 + \sum_{j=1}^{p} \beta_j x_{ij} + \varepsilon_i$$
* $\hat{y}_i$ : valeur prédite pour l'observation $i$ ;
* $\beta_0$ : ordonnée à l'origine (biais) ;
* $\beta_j$ : paramètre appris associé à la $j$-ème variable ;
* $x_{ij}$ : valeur de la $j$-ème variable pour l'individu $i$ ;
* $p$ : nombre total de variables explicatives ;
* $\varepsilon_i$ : terme d'erreur (résidu $y_i - \hat{y}_i$).

### Schéma original (C2)
* *(Insérez ici votre image de schéma via l'icône photo de la barre d'outils)*
* **Légende et choix algorithmique :** Ce schéma illustre la minimisation géométrique de la distance verticale entre les points observés et la droite de régression.

### Contextes et choix (C3)
* **Contexte 1 (Finance / Gestion) :** Prédiction du chiffre d'affaires mensuel d'une entreprise à partir des budgets publicitaires.
* **Contexte 2 (Logistique) :** Estimation des coûts de transport routier selon la distance parcourue.
* **Structure de données attendue :** Tableau structuré de variables numériques continues.
* **Algorithme concurrent :** L'arbre de décision de régression (`DecisionTreeRegressor`).
* **Situation mal adaptée :** Lorsque la relation entre les variables est fortement non-linéaire (courbes, cercles, structures complexes).
* **Justification de la métrique :** Le coefficient de détermination ($R^2$) pour mesurer la proportion de variance expliquée.

### Test de sensibilité et interprétation (C4)
* **Effet de la modification d'hyperparamètre (`fit_intercept=False` vs `True`) :** Forcer l'intercept à zéro modifie l'ajustement global de la droite, augmentant généralement l'erreur si les données ne passent pas par l'origine.
* **Limite d'une seule exécution :** Ce résultat repose sur un seul découpage aléatoire des données (`train_test_split`).

```python
# Test de sensibilité C4 : comparaison avec/sans intercept et diagnostic des résidus
modele_sans_biais = LinearRegression(fit_intercept=False).fit(Xr_ent, yr_ent)
pred_sans_biais = modele_sans_biais.predict(Xr_test)

print(
    "Avec intercept (référence) -> MAE :",
    round(mean_absolute_error(yr_test, predictions), 2),
)
print(
    "Sans intercept (fit_intercept=False) -> MAE :",
    round(mean_absolute_error(yr_test, pred_sans_biais), 2),
)

residus = yr_test - predictions
plt.figure(figsize=(6, 3.5))
plt.scatter(predictions, residus, alpha=0.7)
plt.axhline(0, color="red", linestyle="--")
plt.xlabel("Valeurs prédites")
plt.ylabel("Résidus")
plt.title("Diagnostic des résidus - Régression linéaire (Diabetes)")
plt.show()
```

## 2. Ridge

**Variante C — question spécifique :** Comment évoluent les coefficients quand alpha augmente ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.linear_model import Ridge
modele = make_pipeline(StandardScaler(), Ridge(alpha=12.0)).fit(Xr_ent, yr_ent)
print('MAE Ridge :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
```

### Présentation (C1)
* **Catégorie :** Apprentissage supervisé (Régression linéaire régularisée par pénalité $L_2$).
* **Nature des entrées / sorties :** En entrée, une matrice de variables explicatives $X$ ; en sortie, une variable cible continue $y$.
* **Hypothèse clé :** Les coefficients du modèle ne doivent pas croître excessivement afin de réduire la variance et d'éviter le surajustement en présence de multicolinéarité.
* **Étapes successives de calcul :** (1) Standardisation des données, (2) Ajout d'une pénalité quadratique ($L_2$) sur la norme des coefficients à la fonction de coût, (3) Résolution pour trouver les poids optimaux.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les coefficients de régression $\beta$ et le biais ; l'**hyperparamètre** est le coefficient de régularisation $\alpha$ qui contrôle l'intensité de la pénalité.
* **Équation et définition des symboles :**
$$\min_{\beta} ||y - X\beta||^2_2 + \alpha ||\beta||^2_2$$
* $\min_{\beta}$ : minimisation par rapport au vecteur de coefficients $\beta$ ;
* $||y - X\beta||^2_2$ : somme des erreurs quadratiques (résidus) ;
* $\alpha$ : hyperparamètre de régularisation $L_2$ ;
* $||\beta||^2_2$ : somme des carrés des coefficients ($\sum \beta_j^2$).

### Schéma original (C2)
* *(Insérez ici votre image de schéma via l'icône photo de la barre d'outils)*
* **Légende et choix algorithmique :** Ce schéma illustre la contrainte géométrique en cercle (norme $L_2$) qui contracte continûment les coefficients vers zéro sans jamais les annuler strictement.

### Contextes et choix (C3)
* **Contexte 1 (Finance / Gestion) :** Modélisation et prédiction de prix d'actifs financiers en présence de variables macroéconomiques fortement corrélées.
* **Contexte 2 (Commerce International) :** Estimation des volumes d'exportation en fonction de multiples indicateurs économiques redondants.
* **Structure de données attendue :** Tableau de données tabulaires avec des variables numériques continues (standardisées).
* **Algorithme concurrent :** La régression linéaire classique ou Lasso.
* **Situation mal adaptée :** Lorsque seules quelques variables sont réellement prédictives et que les autres sont du bruit pur (Lasso est meilleur pour éliminer le bruit).
* **Justification de la métrique :** Le $R^2$ ou la MAE pour évaluer la précision de prédiction globale.

### Test de sensibilité et interprétation (C4)
* **Effet de la modification d'hyperparamètre (`alpha`) :** Plus $\alpha$ augmente, plus les coefficients se contractent vers zéro (réponse à la question spécifique : ils diminuent en valeur absolue), réduisant la variance du modèle.
* **Limite d'une seule exécution :** Le résultat dépend du découpage aléatoire des données d'entraînement et de test.

```python
# Test de sensibilité C4 : impact de l'augmentation de alpha sur la somme des coefficients (Réponse à la question spécifique)
modele_faible = make_pipeline(StandardScaler(), Ridge(alpha=0.1)).fit(
    Xr_ent, yr_ent
)
modele_fort = make_pipeline(StandardScaler(), Ridge(alpha=1000.0)).fit(
    Xr_ent, yr_ent
)

print(
    "Somme absolue des coefficients (alpha=0.1) ->",
    round(np.sum(np.abs(modele_faible[-1].coef_)), 2),
)
print(
    "Somme absolue des coefficients (alpha=1000) ->",
    round(np.sum(np.abs(modele_fort[-1].coef_)), 2),
)

# Graphique d'évolution de l'erreur selon alpha
alphas = [0.1, 1.0, 12.0, 100.0, 1000.0]
maes = []
for a in alphas:
  mod = make_pipeline(StandardScaler(), Ridge(alpha=a)).fit(Xr_ent, yr_ent)
  maes.append(mean_absolute_error(yr_test, mod.predict(Xr_test)))

plt.figure(figsize=(6, 3.5))
plt.plot(alphas, maes, marker="o", color="b")
plt.xscale("log")
plt.xlabel("Alpha (log scale)")
plt.ylabel("MAE sur le test")
plt.title("Sensibilité de Ridge face à alpha")
plt.grid(True)
plt.show()
```

## 3. Lasso

**Variante C — question spécifique :** Combien de variables sont annulées pour trois valeurs d’alpha ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.linear_model import Lasso
modele = make_pipeline(StandardScaler(), Lasso(alpha=2.0, max_iter=15000)).fit(Xr_ent, yr_ent)
print('Nombre de coefficients non nuls :', np.count_nonzero(modele[-1].coef_))
print('MAE Lasso :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
```

### Présentation (C1)
* **Catégorie :** Apprentissage supervisé (Régression linéaire régularisée par pénalité $L_1$ et sélection automatique de variables).
* **Nature des entrées / sorties :** En entrée, une matrice de variables explicatives $X$ ; en sortie, une variable cible continue $y$.
* **Hypothèse clé :** Un grand nombre de variables explicatives sont redondantes ou inutiles, et le modèle peut effectuer une sélection automatique en annulant strictement les coefficients non pertinents.
* **Étapes successives de calcul :** (1) Standardisation des données, (2) Optimisation de la fonction de coût pénalisée par la norme $L_1$ (somme des valeurs absolues), (3) Annulation stricte de certains coefficients à zéro.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les coefficients de régression $\beta$ ; l'**hyperparamètre** est le coefficient $\alpha$ qui contrôle la force de la pénalité $L_1$.
* **Équation et définition des symboles :**
$$\min_{\beta} \frac{1}{2n} ||y - X\beta||^2_2 + \alpha ||\beta||_1$$
* $n$ : nombre d'observations dans l'échantillon ;
* $\alpha$ : hyperparamètre de régularisation $L_1$ ;
* $||\beta||_1$ : somme des valeurs absolues des coefficients ($\sum |\beta_j|$).

### Schéma original (C2)
* *(Insérez ici votre image de schéma via l'icône photo de la barre d'outils)*
* **Légende et choix algorithmique :** Ce schéma illustre les contours géométriques de la pénalité $L_1$ en forme de losange dont les sommets pointus touchent les axes, provoquant l'annulation exacte de certains coefficients (parcimonie).

### Contextes et choix (C3)
* **Contexte 1 (Marketing) :** Sélection des caractéristiques clients les plus déterminantes parmi des centaines de variables comportementales.
* **Contexte 2 (Commerce International) :** Identification des barrières tarifaires ou des indicateurs ayant un impact réel et significatif parmi une vaste base de données macroéconomiques.
* **Structure de données attendue :** Données tabulaires avec un grand nombre de variables explicatives numériques.
* **Algorithme concurrent :** Ridge ou ElasticNet.
* **Situation mal adaptée :** Lorsque les variables explicatives sont fortement corrélées par groupes (Lasso en sélectionne arbitrairement une seule et ignore complètement les autres).
* **Justification de la métrique :** Le nombre de coefficients non nuls (mesure de la parcimonie) combiné au $R^2$.

### Test de sensibilité et interprétation (C4)
* **Effet de la modification d'hyperparamètre (`alpha`) :** Un $\alpha$ plus élevé augmente la parcimonie du modèle (réponse à la question spécifique : il y a de plus en plus de variables dont les coefficients sont annulés à zéro).
* **Limite d'une seule exécution :** La stabilité de la sélection des variables dépend de l'échantillon d'entraînement et peut varier légèrement avec un autre découpage.

```python
# Test de sensibilité C4 : impact de alpha sur le nombre de variables annulées à zéro (Réponse à la question spécifique)
alphas_lasso = [0.1, 1.0, 5.0, 10.0]
for a_val in alphas_lasso:
  mod_test = make_pipeline(
      StandardScaler(), Lasso(alpha=a_val, max_iter=15000)
  ).fit(Xr_ent, yr_ent)
  nb_nuls = np.sum(mod_test[-1].coef_ == 0)
  print(
      f"Alpha = {a_val} -> Variables annulées (à 0) : {nb_nuls} sur"
      f" {X_reg.shape[1]} variables"
  )

# Graphique de l'évolution du nombre de coefficients non nuls selon alpha
coefficients_non_nuls = []
for a_val in alphas_lasso:
  mod_test = make_pipeline(
      StandardScaler(), Lasso(alpha=a_val, max_iter=15000)
  ).fit(Xr_ent, yr_ent)
  coefficients_non_nuls.append(np.sum(mod_test[-1].coef_ != 0))

plt.figure(figsize=(6, 3.5))
plt.plot(alphas_lasso, coefficients_non_nuls, marker="s", color="g")
plt.xlabel("Alpha")
plt.ylabel("Nombre de variables conservées (!= 0)")
plt.title("Sensibilité de Lasso : Parcimonie selon alpha")
plt.grid(True)
plt.show()
```

## 4. Régression polynomiale

**Variante C — question spécifique :** Le degré 2 améliore-t-il le test ou seulement l’entraînement ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele = make_pipeline(StandardScaler(), PolynomialFeatures(degree=2, include_bias=False), Ridge(alpha=100.)).fit(Xr_ent, yr_ent)
print('R² polynomial :', round(r2_score(yr_test, modele.predict(Xr_test)), 3))
```

### Présentation (C1)
* **Catégorie :** Apprentissage supervisé (Extension non linéaire de la régression linéaire).
* **Nature des entrées / sorties :** En entrée, des variables explicatives $X$ transformées en puissances et interactions ; en sortie, une variable cible continue $y$.
* **Hypothèse clé :** La relation entre les variables d'entrée et la cible n'est pas linéaire, mais peut être approchée en ajoutant des termes polynomiaux (carrés, cubes) et des interactions.
* **Étapes successive de calcul :** (1) Génération des caractéristiques polynomiales (puissances et combinaisons croisées), (2) Standardisation des nouvelles caractéristiques, (3) Application d'une régression linéaire régularisée (Ridge) pour estimer les poids.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les coefficients de régression associés à chaque terme polynomial ; l'**hyperparamètre** est le degré du polynôme (`degree`) et le coefficient de régularisation (`alpha`).
* **Équation et définition des symboles :**
$$\hat{y} = \beta_0 + \sum_{j=1}^{p} \beta_j x_j + \sum_{j=1}^{p} \beta_{j^2} x_j^2 + \sum_{j < k} \beta_{jk} x_j x_k$$
* $\hat{y}$ : valeur prédite ;
* $\beta_0$ : biais ;
* $x_j^2$ : termes quadratiques ;
* $x_j x_k$ : termes d'interaction entre variables.

### Schéma original (C2)
* *(Insérez ici votre image de schéma via l'icône photo de la barre d'outils)*
* **Légende et choix algorithmique :** Ce schéma illustre comment une courbe polynomiale s'ajuste de manière flexible à un nuage de données non linéaire par rapport à une simple droite rigide.

### Contextes et choix (C3)
* **Contexte 1 (Commerce International / Économie) :** Modélisation de la croissance d'un marché ou de la demande en fonction du temps ou des prix, présentant des inflexions (courbes non linéaires).
* **Contexte 2 (Finance) :** Étude de l'évolution non linéaire des taux d'intérêt ou des rendements selon des facteurs macroéconomiques.
* **Structure de données attendue :** Données tabulaires numériques continues avec des relations non linéaires sous-jacentes.
* **Algorithme concurrent :** Les arbres de décision (`DecisionTreeRegressor`) ou les machines à vecteurs de support (`SVR`).
* **Situation mal adaptée :** Lorsque le nombre de variables de départ est très élevé (le nombre de caractéristiques polynomiales explose combinatoirement, provoquant un surajustement massif).

* **Limite d'une seule exécution :** L'évaluation sur un seul jeu de test ne suffit pas à garantir la robustesse face à l'overfitting.

```python
from sklearn.preprocessing import PolynomialFeatures

# Test de sensibilité C4 : comparaison de l'impact du degré du polynôme (ex: degré 1 vs degré 2) sur l'entraînement et le test
for deg in [1, 2, 3]:
  modele_poly = make_pipeline(
      StandardScaler(),
      PolynomialFeatures(degree=deg, include_bias=False),
      Ridge(alpha=100.0),
  ).fit(Xr_ent, yr_ent)

  r2_ent = modele_poly.score(Xr_ent, yr_ent)
  r2_tst = modele_poly.score(Xr_test, yr_test)
  print(f"Degré {deg} -> R² Entraînement : {r2_ent:.3f} | R² Test : {r2_tst:.3f}")

# Visualisation de la sensibilité du degré sur le score R²
degres = [1, 2, 3]
r2_tests = []
for deg in degres:
  mod = make_pipeline(
      StandardScaler(),
      PolynomialFeatures(degree=deg, include_bias=False),
      Ridge(alpha=100.0),
  ).fit(Xr_ent, yr_ent)
  r2_tests.append(mod.score(Xr_test, yr_test))

plt.figure(figsize=(6, 3.5))
plt.plot(degres, r2_tests, marker="o", color="purple")
plt.xlabel("Degré du polynôme")
plt.ylabel("R² sur le test")
plt.title("Sensibilité de la régression polynomiale selon le degré")
plt.grid(True)
plt.show()
```

### Présentation (C1)
* **Catégorie :** Apprentissage supervisé non paramétrique (Arbre de décision pour la régression).
* **Nature des entrées / sorties :** En entrée, des variables explicatives $X$ numériques ou catégorielles ; en sortie, une variable cible continue $y$.
* **Hypothèse clé :** L'espace des caractéristiques peut être subdivisé de manière récursive en régions rectangulaires homogènes minimisant la variance de la cible.
* **Étapes successives de calcul :** (1) Évaluation de toutes les coupures possibles sur toutes les variables pour trouver celle qui minimise l'erreur quadratique, (2) Division (split) du nœud en deux sous-nœuds, (3) Répétition du processus jusqu'à l'atteinte d'un critère d'arrêt (ex: profondeur maximale).
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les seuils de coupure et les valeurs moyennes prédites dans les feuilles de l'arbre ; les **hyperparamètres** sont `max_depth` (profondeur maximale) et `min_samples_leaf` (nombre minimal d'échantillons par feuille).
* **Équation et définition des symboles :**
$$\min_{j, s} \left[ \sum_{i: x_i \in R_1(j,s)} (y_i - c_1)^2 + \sum_{i: x_i \in R_2(j,s)} (y_i - c_2)^2 \right]$$
* $j$ : variable explicative choisie pour la coupure ;
* $s$ : valeur de seuil de la coupure ;
* $R_1, R_2$ : régions résultant de la division de l'espace ;
* $c_1, c_2$ : moyenne de la cible $y$ dans les régions $R_1$ et $R_2$.

### Présentation (C1)
* **Catégorie :** Apprentissage supervisé par ensemble (Bagging d'arbres de décision pour la régression).
* **Nature des entrées / sorties :** En entrée, des variables explicatives $X$ ; en sortie, une variable cible continue $y$.
* **Hypothèse clé :** En combinant un grand nombre d'arbres de décision entraînés sur des sous-échantillons aléatoires avec des variables sélectionnées au hasard, on réduit considérablement la variance globale sans augmenter le biais.
* **Étapes successives de calcul :** (1) Création de plusieurs échantillons par tirage aléatoire avec remise (bootstrap), (2) Entraînement d'un arbre de décision sur chaque échantillon en ne choisissant qu'un sous-ensemble aléatoire de variables à chaque division, (3) Moyennage des prédictions de tous les arbres pour obtenir le résultat final.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont l'ensemble des seuils et des structures de tous les arbres individuels de la forêt ; les **hyperparamètres** sont `n_estimators` (nombre d'arbres) et `max_features` (nombre de variables testées par division).
* **Équation et définition des symboles :**
$$\hat{y} = \frac{1}{B} \sum_{b=1}^{B} T_b(x)$$
* $B$ : nombre total d'arbres dans la forêt (`n_estimators`) ;
* $T_b(x)$ : prédiction renvoyée par l'arbre individuel $b$ pour l'entrée $x$.

### Présentation (C1)
* **Catégorie :** Apprentissage supervisé par ensemble (Boosting séquentiel d'arbres de décision).
* **Nature des entrées / sorties :** En entrée, des variables explicatives $X$ ; en sortie, une variable cible continue $y$.
* **Hypothèse clé :** Les performances globales sont améliorées en construisant les arbres de manière séquentielle, chaque nouvel arbre venant corriger les erreurs (résidus) commises par les arbres précédents.
* **Étapes successives de calcul :** (1) Initialisation avec une prédiction constante, (2) Calcul des gradients (résidus) de la fonction de coût, (3) Ajustement d'un arbre de régression pour prédire ces gradients, (4) Mise à jour du modèle global par sommation pondérée.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les poids et les structures des arbres successifs ; les **hyperparamètres** sont `max_iter` (nombre d'itérations/arbres) et `max_leaf_nodes` (nombre maximum de nœuds terminaux par arbre).
* **Équation et définition des symboles :**
$$F_m(x) = F_{m-1}(x) + \gamma_m h_m(x)$$
* $F_m(x)$ : modèle cumulé à l'étape $m$ ;
* $h_m(x)$ : arbre correcteur entraîné sur les résidus à l'étape $m$ ;
* $\gamma_m$ : taux d'apprentissage ou poids de l'arbre.

### Présentation (C1)
* **Catégorie :** Apprentissage supervisé (Modèle de classification linéaire).
* **Nature des entrées / sorties :** En entrée, une matrice de variables explicatives $X$ ; en sortie, une variable cible discrète binaire $y$ (ex: 0 ou 1).
* **Hypothèse clé :** La probabilité d'appartenance à une classe peut être modélisée en appliquant la fonction logistique (sigmoïde) à une combinaison linéaire des variables explicatives.
* **Étapes successives de calcul :** (1) Standardisation des données, (2) Calcul d'une combinaison linéaire des caractéristiques, (3) Application de la fonction sigmoïde pour transformer le score en probabilité entre 0 et 1, (4) Optimisation par maximum de vraisemblance.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les coefficients $\beta$ et le biais ; l'**hyperparamètre** est le paramètre de régularisation inverse `C` (qui contrôle la force de la pénalité).
* **Équation et définition des symboles :**
$$P(y=1|X) = \sigma(\beta_0 + \sum_{j=1}^{p} \beta_j x_j) = \frac{1}{1 + e^{-(\beta_0 + \sum \beta_j x_j)}}$$
* $P(y=1|X)$ : probabilité estimée que l'observation appartienne à la classe positive ;
* $\sigma(\cdot)$ : fonction sigmoïde ;
* $\beta_j$ : coefficients appris.

* **Légende et choix algorithmique :** Ce schéma illustre la transformation linéaire d'un espace de caractéristiques en une probabilité via la courbe en "S" de la sigmoïde, définissant une frontière de décision linéaire.

* **Situation mal adaptée :** Lorsque les classes s'imbriquent de manière circulaire ou très non-linéaire (nécessite des noyaux ou des modèles non-linéaires).
* **Justification de la métrique :** Le score F1 (`f1_score`) ou la ROC-AUC pour évaluer la qualité de la classification binaire.

### Test de sensibilité et interprétation (C4)
* **Effet de la modification d'hyperparamètre (`C`) :** Un `C` élevé réduit la régularisation (frontière de décision plus stricte et ajustée au plus près des données), tandis qu'un `C` faible augmente la régularisation (frontière plus souple et stable, réponse à la question spécifique sur l'effet de C sur les probabilités et la frontière).
* **Limite d'une seule exécution :** L'évaluation dépend du découpage stratifié de l'échantillon d'entraînement et de test.

### Présentation (C1)
* **Catégorie :** Apprentissage supervisé non paramétrique (Classification paresseuse / *lazy learning*).
* **Nature des entrées / sorties :** En entrée, une matrice de variables explicatives $X$ ; en sortie, une variable cible discrète $y$ (classes).
* **Hypothèse clé :** Des observations proches dans l'espace des caractéristiques appartiennent généralement à la même classe.

* $p$ : nombre de dimensions (caractéristiques).

* **Hypothèse clé :** Hypothèse "naïve" d'indépendance conditionnelle entre toutes les variables explicatives sachant la classe cible.

* $P(y|X)$ : probabilité a posteriori de la classe $y$ sachant les entrées ;
* $P(y)$ : probabilité a priori de la classe ;
* $P(x_j|y)$ : probabilité conditionnelle de la variable $x_j$ sachant la classe.

* $K$ : nombre total de classes.

$$\hat{y} = \text{mode}\{T_1(x), T_2(x), \dots, T_B(x)\}$$

* $T_b(x)$ : prédiction de classe renvoyée par l'arbre $b$.

* **Hypothèse clé :** Les classes peuvent être séparées par un hyperplan maximisant la distance (marge) entre les points les plus proches (vecteurs de support) de chaque classe.
* **Étapes successives de calcul :** (1) Standardisation des données, (2) Résolution d'un problème d'optimisation sous contrainte pour trouver l'hyperplan optimal maximisant la marge, (3) Utilisation optionnelle d'une fonction noyau (kernel) pour projeter les données en dimension supérieure en cas de non-linéarité.

* $C$ : paramètre de pénalité des erreurs de classification (compromis marge/marge molle) ;
* $\xi_i$ : variables d'écart (slack variables) mesurant la violation de la marge.

* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les structures et poids des arbres successifs ; les **hyperparamètres** sont `max_iter` (nombre d'itérations) et `max_leaf_nodes` (complexité des arbres).
* **Équation et définition des symboles :**
$$F_m(x) = F_{m-1}(x) + \gamma_m h_m(x)$$
* $F_m(x)$ : modèle additif cumulé à l'étape $m$ ;
* $h_m(x)$ : arbre correcteur entraîné sur le gradient à l'étape $m$.

* **Nature des entrées / sorties :** En entrée, une matrice de variables explicatives $X$ sans étiquette cible ; en sortie, un ensemble de clusters (groupes d'observations) indicés.
* **Hypothèse clé :** Les données peuvent être regroupées en $K$ groupes distincts autour de centroïdes, minimisant la variance intra-groupe.
* **Étapes successives de calcul :** (1) Standardisation des données, (2) Initialisation aléatoire de $K$ centroïdes, (3) Affectation de chaque observation au centroïde le plus proche, (4) Recalcul de la position des centroïdes comme moyenne des points assignés, (5) Répétition jusqu'à convergence.

* **Équation et définition des symboles :**
$$J = \sum_{k=1}^{K} \sum_{i \in S_k} ||x_i - \mu_k||^2$$
* $K$ : nombre de clusters à former ;
* $S_k$ : ensemble des points appartenant au cluster $k$ ;
* $\mu_k$ : centroïde (moyenne) du cluster $k$.

* **Étapes successives de calcul :** (1) Standardisation des données, (2) Pour chaque point, calcul du nombre de voisins dans un rayon $\epsilon$ (`eps`), (3) Identification des points centraux (*core points*) ayant au moins `min_samples` voisins, (4) Propagation et extension des clusters par connexité de densité, (5) Marquage des points isolés comme bruit (étiquette -1).

* `min_samples` : nombre minimal de points requis dans le voisinage $\epsilon$ pour former un noyau dense.

### Présentation (C1)
* **Catégorie :** Apprentissage supervisé à vaste marge (Régression par machines à vecteurs de support).
* **Nature des entrées / sorties :** En entrée, une matrice de variables explicatives $X$ ; en sortie, une variable cible continue $y$.
* **Hypothèse clé :** Trouver une fonction de prédiction qui s'écarte de la cible réelle $y$ d'au plus une tolérance $\epsilon$, tout en maintenant le modèle aussi plat et régulier que possible.
* **Étapes successives de calcul :** (1) Standardisation des données, (2) Résolution d'un problème d'optimisation sous contrainte pour trouver un tube de tolérance insensible $\epsilon$ maximisant la marge, (3) Utilisation optionnelle d'un noyau (kernel) pour projeter les données en dimension supérieure en cas de non-linéarité.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les poids de l'hyperplan et les vecteurs de support ; les **hyperparamètres** sont `C` (pénalité des erreurs hors tube) et `epsilon` ($\epsilon$, largeur de la zone insensible).
* **Équation et définition des symboles :**
$$\min_{w, b} \frac{1}{2} ||w||^2 + C \sum_{i=1}^{n} (\xi_i + \xi_i^*)$$
* $w$ : vecteur normal régissant la pente de la fonction de régression ;
* $C$ : paramètre de régulation du compromis entre la platitude et l'acceptation des écarts ;

### Présentation (C1)
* **Catégorie :** Apprentissage supervisé non paramétrique (Régression par les $k$ plus proches voisins).
* **Nature des entrées / sorties :** En entrée, une matrice de variables explicatives $X$ ; en sortie, une variable cible continue $y$.
* **Hypothèse clé :** La valeur cible d'une nouvelle observation peut être estimée en calculant la moyenne (ou la moyenne pondérée) des valeurs cibles de ses $k$ voisins les plus proches dans l'espace des caractéristiques.
* **Étapes successives de calcul :** (1) Standardisation des données, (2) Calcul des distances géométriques (ex: euclidienne) entre le point de test et l'ensemble des points d'entraînement, (3) Sélection des $k$ voisins les plus proches, (4) Calcul de la prédiction par moyenne arithmétique ou pondérée par l'inverse de la distance.
* **Paramètre appris vs Hyperparamètre :** Aucun paramètre n'est explicitement appris par optimisation (stockage des données / apprentissage paresseux) ; les **hyperparamètres** sont `n_neighbors` ($k$) et le mode de pondération (`weights`).
* **Équation et définition des symboles :**
$$\hat{y} = \frac{1}{k} \sum_{i \in N_k(x)} y_i$$
* $k$ : nombre de voisins pris en compte (`n_neighbors`) ;
* $N_k(x)$ : ensemble des $k$ observations les plus proches de l'entrée $x$ ;
* $y_i$ : valeur cible de l'observation voisine $i$.

* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les moyennes et les matrices de covariance de chaque classe ; l'**hyperparamètre** principal est `reg_param` (régularisation de la matrice de covariance).
* **Équation et définition des symboles :**
$$\delta_k(x) = -\frac{1}{2} \ln |\Sigma_k| - \frac{1}{2} (x - \mu_k)^T \Sigma_k^{-1} (x - \mu_k) + \ln P(y=k)$$
* $\Sigma_k$ : matrice de covariance propre à la classe $k$ ;
* $\mu_k$ : vecteur des moyennes de la classe $k$.

```python
# Q-learning manuel — environnement CliffWalking (Gymnasium)
import gymnasium as gym
env = gym.make('CliffWalking-v1')
q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(GRAINE)
for episode in range(1100):
    etat, _ = env.reset(seed=GRAINE + episode)
    epsilon = max(.08, .8 * (1 - episode / 900))
    for pas in range(120):
        action = int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(q_table[etat]))
        suivant, recompense, termine, tronque, _ = env.step(action)
        cible = recompense + (0 if termine else .95 * np.max(q_table[suivant]))
        q_table[etat, action] += .15 * (cible - q_table[etat, action])
        etat = suivant
        if termine or tronque: break
print('Table Q :', q_table.shape, '| valeur moyenne :', round(q_table.mean(), 2))
env.close()
```

* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les poids de l'hyperplan séparateur et le biais ; les **hyperparamètres** sont `loss` (fonction de perte, ex: `log_loss` ou `hinge`), `penalty` (régularisation) et `alpha` (force de pénalité).
* **Équation et définition des symboles :**
$$w^{(t+1)} = w^{(t)} - \eta_t \nabla L(w; x_i, y_i)$$
* $w$ : vecteur normal de l'hyperplan ;

## 28. SARSA

**Variante C — question spécifique :** SARSA privilégie-t-il une trajectoire plus sûre sous exploration ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

```python
# SARSA manuel — même environnement, politique epsilon-greedy suivie
import gymnasium as gym
env = gym.make('CliffWalking-v0')
q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(GRAINE)
def choisir_action(etat, epsilon):
    return int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(q_table[etat]))
for episode in range(1100):
    etat, _ = env.reset(seed=GRAINE + episode)
    epsilon = max(.08, .8 * (1 - episode / 900))
    action = choisir_action(etat, epsilon)
    for pas in range(120):
        suivant, recompense, termine, tronque, _ = env.step(action)
        action_suivante = choisir_action(suivant, epsilon)
        cible = recompense + (0 if termine else .95 * q_table[suivant, action_suivante])
        q_table[etat, action] += .15 * (cible - q_table[etat, action])
        etat, action = suivant, action_suivante
        if termine or tronque: break
print('Table Q SARSA :', q_table.shape, '| valeur moyenne :', round(q_table.mean(), 2))
env.close()
```

### Présentation (C1)
* **Catégorie :** Apprentissage supervisé par ensemble (Boosting adaptatif séquentiel).
* **Nature des entrées / sorties :** En entrée, une matrice de variables explicatives $X$ ; en sortie, une variable cible discrète $y$ (classes).
* **Hypothèse clé :** L'association séquentielle de classifieurs faibles (*weak learners*, généralement des arbres de décision à un seul niveau ou *stumps*) permet de construire un modèle global robuste en accordant plus de poids aux observations mal classées par les étapes précédentes.
* **Étapes successives de calcul :** (1) Initialisation de poids uniformes pour toutes les observations, (2) Entraînement d'un classifieur faible sur l'échantillon pondéré, (3) Calcul du taux d'erreur et attribution d'un poids de vote au modèle, (4) Augmentation des poids des observations mal classées pour l'itération suivante, (5) Combinaison linéaire pondérée des votes de tous les estimateurs.
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les poids de vote des estimateurs et les seuils de coupure des arbres faibles ; les **hyperparamètres** sont `n_estimators` (nombre d'arbres séquentiels) et `learning_rate` (taux d'apprentissage).
* **Équation et définition des symboles :**
$$H(x) = \text{sign} \left( \sum_{m=1}^{M} \alpha_m h_m(x) \right)$$
* $M$ : nombre total d'estimateurs (`n_estimators`) ;
* $h_m(x)$ : prédiction du classifieur faible $m$ ;
* $\alpha_m$ : coefficient de pondération du classifieur $m$ basé sur son taux d'erreur.

* **Légende et choix algorithmique :** Ce schéma illustre la correction itérative des erreurs où chaque modèle faible se concentre prioritairement sur les échantillons pénalisés (mal classés) de l'étape précédente.

### Contextes et choix (C3)
* **Contexte 1 (Finance) :** Détection de fraudes bancaires ou de risques de défaut de paiement hautement déséquilibrés.

```python
# Cibles DQN : démonstration du calcul de Bellman, SANS réseau neuronal
recompenses = np.array([-1., -100., 0.])
termine = np.array([False, True, False])
meilleure_valeur_suivante = np.array([-3., -2., 5.])
gamma = .9
cibles = recompenses + gamma * (~termine) * meilleure_valeur_suivante
print('Cibles de Bellman calculées :', np.round(cibles, 2))
print('Ce calcul ne constitue pas un entraînement DQN.')
```

### Présentation (C1)
* **Catégorie :** Apprentissage supervisé par ensemble (Boosting adaptatif séquentiel pour la régression).
* **Nature des entrées / sorties :** En entrée, une matrice de variables explicatives $X$ ; en sortie, une variable cible continue $y$.
* **Hypothèse clé :** L'association séquentielle d'estimateurs de régression faibles permet d'ajuster une cible continue en accordant un poids plus élevé aux observations présentant les erreurs résiduelles les plus importantes lors des itérations précédentes.
* **Étapes successives de calcul :** (1) Initialisation de poids uniformes pour toutes les observations, (2) Entraînement d'un régresseur faible sur l'échantillon pondéré, (3) Calcul des erreurs et mise à jour des poids des observations mal prédites, (4) Combinaison pondérée finale des prédictions des différents estimateurs (généralement par moyenne pondérée ou médiane).
* **Paramètre appris vs Hyperparamètre :** Les **paramètres appris** sont les poids des estimateurs successifs et les structures des modèles faibles ; les **hyperparamètres** sont `n_estimators` (nombre d'itérations) et `learning_rate` (taux d'apprentissage).
* **Équation et définition des symboles :**
$$\hat{y} = \sum_{m=1}^{M} w_m h_m(x)$$
* $M$ : nombre total d'estimateurs (`n_estimators`) ;
* $h_m(x)$ : prédiction du régresseur faible $m$ ;
* $w_m$ : coefficient de pondération du modèle $m$.

* **Structure de données attendue :** Données tabulaires structurées (numériques et/ou catégorielles encodées).

* **Justification de la métrique :** La MAE (Mean Absolute Error) ou le RMSE pour évaluer l'écart de prédiction continu.

### Test de sensibilité et interprétation (C4)
* **Effet de la modification d'hyperparamètre (`learning_rate`) :** Un taux d'apprentissage plus élevé accélère la correction des erreurs mais risque d'introduire de l'instabilité (réponse à la question spécifique : un `learning_rate` faible combiné à un `n_estimators` élevé améliore souvent la généralisation au détriment du temps de calcul).
* **Limite d'une seule exécution :** L'évaluation sur un unique ensemble de test ne suffit pas à garantir la robustesse absolue du modèle face aux variations d'échantillonnage.
